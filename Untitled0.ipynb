```python
import pandas as pd
import zipfile
import os

# 1. Caminho padrão do Colab
zip_path = '/content/archive.zip' # Altere 'archive.zip' se o seu arquivo tiver outro nome
extract_dir = '/content/data'

# Cria a pasta data se não existir
if not os.path.exists(extract_dir):
    os.makedirs(extract_dir)

# Extrai os arquivos
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print("Arquivos extraídos:", os.listdir(extract_dir))

# 2. Carregar o CSV no Pandas
csv_file = [f for f in os.listdir(extract_dir) if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(extract_dir, csv_file))

print(f"\nDataset carregado! Tamanho: {df.shape[0]} linhas e {df.shape[1]} colunas.")
display(df.head(3))
```

```python
import numpy as np

# 1. Converter strings para objetos datetime do Pandas
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])

# 2. Calcular o tempo de espera em dias
# Normalizamos para a meia-noite para que a diferença em dias seja exata
df['WaitTime'] = (df['AppointmentDay'].dt.normalize() - df['ScheduledDay'].dt.normalize()).dt.days

# 3. Corrigindo nomes de colunas com erros de digitação
df = df.rename(columns={'Hipertension': 'Hypertension', 'Handcap': 'Handicap', 'No-show': 'NoShow'})

# 4. Analisar os resultados da nova coluna
print("Resumo estatístico do Tempo de Espera (em dias):")
display(df['WaitTime'].describe())

# 5. Identificando anomalias para citar no artigo
anomalias_tempo = df[df['WaitTime'] < 0]
print(f"\nAtenção: Encontramos {len(anomalias_tempo)} consultas com tempo de espera NEGATIVO.")
print("Isso deve ser limpo no próximo passo do pré-processamento.")
```
